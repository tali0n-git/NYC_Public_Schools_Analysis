```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

mta_time_1 = pd.read_csv('../data/MTA_Subway_Terminal_On-Time_Performance_2015-2019.csv', low_memory=False)
mta_time_2 = pd.read_csv('../data/MTA_Subway_Terminal_On-Time_Performance_2020-2024.csv', low_memory=False)

mta_time_1.head()
```
