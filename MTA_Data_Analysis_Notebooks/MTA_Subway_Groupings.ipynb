# MTA Subway Line Groupings

## Goal:
- Group the subway stations into their respective train lines: 

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

stations = pd.read_csv('../cleaned_data/MTA_Subway_Stations_With_Census_Tracts_2025.csv', low_memory=False)

stations.shape
```

```python
stations[stations['Daytime Routes'].str.contains('J', na=False)]
```

```python
stations[stations['Daytime Routes'].str.contains('Z', na=False)]
```

# Note:
- Z always appears with J (it is the express train w.r.t. the J train)

```python
# Split rows into multiple rows based on 'Daytime Routes' column
stations_expanded = stations.assign(**{
    'Daytime Routes': stations['Daytime Routes'].str.split(' ')
}).explode('Daytime Routes')
stations_expanded.head()
```

```python
stations_expanded.shape
```

```python
stations_expanded[stations_expanded['Daytime Routes'] == 'J']
```

```python
stations_expanded[stations_expanded['Daytime Routes'] == 'Z']
```
