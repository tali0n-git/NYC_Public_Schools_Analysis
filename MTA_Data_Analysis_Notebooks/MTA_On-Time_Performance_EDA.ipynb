# Note:
- From 2015-2019 (mta_time_1) there is no J line distinction, only JZ
- From 2020 onward (mta_time_2) there ARE line distinctions: J and JZ lines
- FS and GS line distinction are also introduced in 2020
- ALSO, H line introduced in 2020 (H = legacy label for 'Rockaway Park Shuttle')

## Converting S Fkln -> FS and S 42nd to GS:
- And S Rock -> RS
- And H -> RS


```python
mta_time_1['line'] = mta_time_1['line'].apply(lambda line: 'FS' if line == 'S Fkln' else 'GS' if line == 'S 42nd' else 'RS' if line == 'S Rock' else line)
mta_time_2['line'] = mta_time_2['line'].apply(lambda line: 'FS' if line == 'S Fkln' else 'GS' if line == 'S 42nd' else 'RS' if (line == 'S Rock' or line == 'H') else line)
```

```python
# Combining the two datasets:
mta_time_df = pd.concat([mta_time_1, mta_time_2], ignore_index=True)

mta_time_df.shape
```

## Converting numerical data into datetimes, ints, and floats:

```python
mta_time_df.dtypes
```

```python
# Datetime conversion:
mta_time_df['month'] = pd.to_datetime(mta_time_df['month'], format='%Y-%m-%d')
```

```python
# Int conversion:
mta_time_df['num_on_time_trips'] = mta_time_df['num_on_time_trips'].apply(lambda x: int(x.replace(',', '')))
mta_time_df['num_sched_trips'] = mta_time_df['num_sched_trips'].apply(lambda x: int(x.replace(',', '')))
```

```python
# Float conversion:
mta_time_df['terminal_on_time_performance'] = mta_time_df['terminal_on_time_performance'].apply(lambda x: float(x.replace('%', '')))
```

## Creating 'Daytime Routes' column, changing JZ -> Z line data
- NO exclusively J data between 2015 and 2019!

```python
mta_time_df['Daytime Routes'] = mta_time_df['line'].apply(lambda line: 'Z' if line == 'JZ' else line)

mta_time_df.sample(10)
```

```python
mta_time_df.to_json('../cleaned_data/MTA_Subway_Performance.json', orient='index')
```
