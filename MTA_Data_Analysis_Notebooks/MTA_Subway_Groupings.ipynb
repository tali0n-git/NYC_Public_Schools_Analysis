## Assigning 'S' stations one of 3 shuttle designations: RS, FS, or GS:

```python
stations_expanded[stations_expanded['Daytime Routes'] == 'S']
```

```python
def update_daytime_routes(row):
    if 'S' not in row['Daytime Routes']:
        return row['Daytime Routes']
    
    if row['Borough'] == 'M':
        return 'GS'
    elif row['Borough'] == 'Bk':
        return 'FS'
    elif row['Borough'] == 'Q':
        return 'RS'
    else:
        return row['Daytime Routes']

stations_expanded['Daytime Routes'] = stations_expanded.apply(update_daytime_routes, axis=1)
```

## Process:
- Create a dictionary of subway lines
- Populate with: 
* station_expanded[[Census Tract Whole', 'Borough', 'Stop Name']] 
**AND** 
* mta_metrics[['month', 'terminal_on_time_performance']]     (mapped on='Daytime Routes')

- Export into new file

```python
stations_grouped = stations_expanded.groupby('Daytime Routes').agg(
    num_stations = ('Daytime Routes', 'count'),
    station_latitudes = ('GTFS Latitude', lambda x: x.tolist()),
    station_longitudes = ('GTFS Longitude', lambda x: x.tolist()),
    station_census_tracts = ('Census Tract Whole', lambda x: x.tolist()),
    station_zip_codes = ('zip_code', lambda x: x.tolist()),
    station_boroughs = ('Borough', lambda x: x.tolist()),
    stop_names = ('Stop Name', lambda x: x.tolist())
)

stations_grouped
```

```python
stations_list = stations_grouped.index.to_list()
stations_list
```

```python
stations_grouped.loc['Z']
```

```python
def grouped_to_dict(grouped_df):
    """
    Convert a grouped DataFrame into a dictionary where:
    - Key: the group name (station/route)
    - Value: dictionary containing all grouped data
    """
    result = {}
    for station, row in grouped_df.iterrows():
        result[station] = row.to_dict()
    return result

subway_dict = grouped_to_dict(stations_grouped)
subway_dict['J']
```

```python
for indx, row in mta_metrics.iterrows():
    if row['line'] in subway_dict:
        if 'mta_performance_data' not in subway_dict[row['line']]:
            subway_dict[row['line']]['mta_performance_data'] = []
        subway_dict[row['line']]['mta_performance_data'].append([row['month'], row['terminal_on_time_performance'], row['day_type']])
```

```python
subway_dict['A']['mta_performance_data'][:5]
```

```python
subway_df = pd.DataFrame.from_dict(subway_dict, orient='index')

subway_df.tail()
```
